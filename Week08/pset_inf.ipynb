**Problem 4**

```python


ames = pd.read_csv('./data/ames_prices.csv')
ames['central_air'] = (ames['Central.Air'] == 'Y').astype(int)
x4 = ames['central_air']
p_hat = x4.mean()

np.random.seed(5030)
R = 5_000
boot4 = []
for t in range(R):
    idx = np.random.randint(0, len(x4), len(x4))
    boot4.append(x4.iloc[idx].mean())
boot4 = np.array(boot4)

se4 = np.std(boot4, ddof=1)
z4 = (p_hat - 0.90) / se4
p4 = 2 * norm.cdf(-np.abs(z4))
print(f'Seed = 5030, R = {R}')
print(f'p_hat = {p_hat:.4f}, SE = {se4:.4f}, z = {z4:.3f}, p = {p4:.3g}')
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    lo, hi = p_hat - c*se4, p_hat + c*se4
    print(f'{conf}% CI: ({lo:.4f}, {hi:.4f}) | reject H0: {p4 < alpha} | CI excludes 0.90: {not (lo <= 0.90 <= hi)}')
```

- Estimate: A p hat of 0.9331 means that out of the houses sample in Ames, about 93.31% have central ac. SE=0.0045, seed=5030, R=5000.  
- Interval: The 90% confidence interval estimates that the range of homes with central ac is 92.56% to 94.06%. 90% confident that the actual ac pct is somewhere in this range. 95% confident that actual ac pct is in 92.42%, 94.2%. 99% confident that actual ac pct is in 92.14%, 94.48%. 
- Test: Since the z-val = 7.290 and the p-val = 3.11e-13 (much less than 0.05, 0.01, 0.10), we reject the null hypothesis for all 3 significance levels. The null hypothesis was that the ac proportion was 90%, which is a 2 way p, and in this case the proportion was higher.

**Problem 5**

```python
area5 = np.arcsinh(ames['area'])
price5 = np.arcsinh(ames['price'])
r_hat = np.corrcoef(area5, price5)[0, 1]

np.random.seed(5030)
R = 5_000
boot5 = []
for t in range(R):
    idx = np.random.randint(0, len(ames), len(ames))  # same idx for both columns keeps rows paired
    boot5.append(np.corrcoef(area5.iloc[idx], price5.iloc[idx])[0, 1])
boot5 = np.array(boot5)

se5 = np.std(boot5, ddof=1)
z5 = (r_hat - 0) / se5
p5 = 2 * norm.cdf(-np.abs(z5))
print(f'Seed = 5030, R = {R}')
print(f'r_hat = {r_hat:.4f}, SE = {se5:.4f}, z = {z5:.3f}, p = {p5:.3g}')
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    lo, hi = r_hat - c*se5, r_hat + c*se5
    print(f'{conf}% CI: ({lo:.4f}, {hi:.4f}) | reject H0: {p5 < alpha} | CI excludes 0: {not (lo <= 0 <= hi)}')
```

- Estimate: An r hat (pearson correlation) of 0.7232 is positive, indicating that in this sample, there is a strong positive linear correlation between arcsin(area) and arcsin(price). SE=0.0100, seed=5030, R=5000.
- Interval: We are 90% confident that the population correlation falls between 0.7067 and 0.7397, 95% confident that it falls between 0.7035 and 0.7428, and 99% confident that it falls between 0.6973 and 0.7490. All 3 CIs match. 
- Test: We reject the null hypothesis (p=0) at all 3 significance levels. P is close to 0 and z=72.032. There is statistically significant evidence that arcsin(area) and arcsin(price) have a linear positive correlation.
- If area and price were resampled separately, the values for each would not match with one another, leading to inaccurate metrics (low correlation metrics). You could have large houses paired with low prices, or small houses paired with high prices. 

**Problem 6**

```python
metabric = pd.read_csv('./data/metabric.csv')
x6 = metabric['Overall Survival (Months)'].dropna()
mean6 = x6.mean()

np.random.seed(5030)
R = 5_000
boot6 = []
for t in range(R):
    idx = np.random.randint(0, len(x6), len(x6))
    boot6.append(x6.iloc[idx].mean())
boot6 = np.array(boot6)

se6 = np.std(boot6, ddof=1)
z6 = (mean6 - 120) / se6
p6 = 2 * norm.cdf(-np.abs(z6))
print(f'Seed = 5030, R = {R}')
print(f'mean = {mean6:.2f} months ({mean6/12:.2f} years), SE = {se6:.2f} months, z = {z6:.3f}, p = {p6:.3g}')
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    lo, hi = mean6 - c*se6, mean6 + c*se6
    print(f'{conf}% CI: ({lo:.2f}, {hi:.2f}) months = ({lo/12:.2f}, {hi/12:.2f}) years | '
          f'reject H0: {p6 < alpha} | CI excludes 120: {not (lo <= 120 <= hi)}')
```

- Estimate: The mean is 129.09 months with an SE of 2.11 months. 
- Interval: We are 90% confident that the mean falls within 125.62 months and 132.57 months, 95% confident that the mean falls within 124.95, 133.24 months, 99% confident in 123.65, 134.54 months. 
- Test: We reject the null hypothesis (mean = 120 months) for all 3 significance levels. p=1.7e-05, z=4.3. The mean recorded time is greater than 120 months. All 3 CIs exclude 120, match. 
- Right censored in this case means that since there are still living patients, we do not yet know their real time survived. The distinction is that because of this, the data does not reveal time survived, but rather the mean recorded time post-appointment. This mean is an underestimation. To get mean survival time post-appointment, you would need to repeat the calculation of the metrics once all the patients in the study had passed away (very morbid).

**Problem 7**

```python
nhanes = pd.read_csv('./data/nhanes_data_17_18.csv')
adults = nhanes[(nhanes['AgeInYearsAtScreening'] >= 18) & nhanes['SleepHoursWeekdaysOrWorkdays'].notna()]
x7 = adults['SleepHoursWeekdaysOrWorkdays']
mean7 = x7.mean()

np.random.seed(5030)
R = 5_000
boot7 = []
for t in range(R):
    idx = np.random.randint(0, len(x7), len(x7))
    boot7.append(x7.iloc[idx].mean())
boot7 = np.array(boot7)

se7 = np.std(boot7, ddof=1)
z7 = (mean7 - 8) / se7
p7 = 2 * norm.cdf(-np.abs(z7))
print(f'Seed = 5030, R = {R}, n = {len(x7)}')
print(f'mean = {mean7:.3f} hours, SE = {se7:.4f}, z = {z7:.3f}, p = {p7:.3g}')
for conf, alpha in [(90, .10), (95, .05), (99, .01)]:
    c = norm.ppf(1 - alpha / 2)
    lo, hi = mean7 - c*se7, mean7 + c*se7
    print(f'{conf}% CI: ({lo:.3f}, {hi:.3f}) | reject H0: {p7 < alpha} | CI excludes 8: {not (lo <= 8 <= hi)}')
```

- Estimate: The estimated mean 7.621 hours with an SE of 0.0225 hours. 
- Interval: We are 90% confident that the mean is between 7.584, 7.658 hours, 95% confident that the mean is between 7.577, 7.665 hours, 99% confident 7.563, 7.679 hours. 
- Test: We reject the null hypothesis (mean = 8) at all 3 levels. z = -16.83 and p=1.61e-63. All 3 CIs exclude 8, these match. 
- Statistical and practical significance do not tell the same story. As samples get very large, small differences are highly statistically significant. Practically, these differences are not very large, as they only amount to minutes.
